```python
k_values = [3, 5, 7, 9, 11]

knn_results = []

for k in k_values:
    knn_model = KNeighborsClassifier(n_neighbors=k)
    
    knn_model.fit(X_train_preprocessed, Y_train)
    
    predictions = knn_model.predict(X_test_preprocessed)
    
    knn_results.append({
        "K": k,
        "Accuracy": accuracy_score(Y_test, predictions),
        "Precision": precision_score(Y_test, predictions),
        "Recall": recall_score(Y_test, predictions),
        "F1 Score": f1_score(Y_test, predictions)
    })

knn_tuning_results = pd.DataFrame(knn_results)

print(knn_tuning_results)
```

```python


sns.lineplot(
    data=knn_tuning_results,
    x="K",
    y="F1 Score",
    marker="o"
)

plt.xlabel("Number of Neighbors (K)")
plt.ylabel("F1 Score")
plt.title("KNN Hyperparameter Tuning")
plt.ylim(0, 0.55)
plt.show()
```

```python

from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

# Train the Decision Tree
dt_model = DecisionTreeClassifier(
    max_depth=5,
    random_state=42,
    class_weight="balanced"
)

dt_model.fit(X_train_preprocessed, Y_train)

# Predict
y_pred_dt = dt_model.predict(X_test_preprocessed)

# Evaluate
print("Decision Tree Results")
print("-" * 30)
print("Accuracy :", accuracy_score(Y_test, y_pred_dt))
print("Precision:", precision_score(Y_test, y_pred_dt, zero_division=0))
print("Recall   :", recall_score(Y_test, y_pred_dt, zero_division=0))
print("F1 Score :", f1_score(Y_test, y_pred_dt, zero_division=0))
print("Confusion Matrix:\n", confusion_matrix(Y_test, y_pred_dt))
```

```python

from sklearn.ensemble import RandomForestClassifier

# Train the Random Forest
rf_model = RandomForestClassifier(
    n_estimators=100,
    max_depth=10,
    random_state=42,
    class_weight="balanced",
    n_jobs=-1
)

rf_model.fit(X_train_preprocessed, Y_train)

# Predict
y_pred_rf = rf_model.predict(X_test_preprocessed)

# Evaluate
print("Random Forest Results")
print("-" * 30)
print("Accuracy :", accuracy_score(Y_test, y_pred_rf))
print("Precision:", precision_score(Y_test, y_pred_rf, zero_division=0))
print("Recall   :", recall_score(Y_test, y_pred_rf, zero_division=0))
print("F1 Score :", f1_score(Y_test, y_pred_rf, zero_division=0))
print("Confusion Matrix:\n", confusion_matrix(Y_test, y_pred_rf))
```

```python

# Random Forest Feature Importance

feature_names = numeric_cols + list(
    encoder.get_feature_names_out(categorical_cols)
)

feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Importance": rf_model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

print(feature_importance.head(15))

plt.figure(figsize=(10, 6))
sns.barplot(
    data=feature_importance.head(15),
    x="Importance",
    y="Feature"
)

plt.title("Top 15 Random Forest Features")
plt.xlabel("Feature Importance")
plt.ylabel("Feature")
plt.tight_layout()
plt.show()
```

```python

# Random Forest without PageValues

numeric_cols_without_pagevalues = [
    col for col in numeric_cols if col != "PageValues"
]

# Scale numeric features without PageValues
X_train_num_no_pv = scaler.fit_transform(
    X_train[numeric_cols_without_pagevalues]
)

X_test_num_no_pv = scaler.transform(
    X_test[numeric_cols_without_pagevalues]
)

# Encode categorical features using the existing encoder
X_train_no_pv = np.hstack([
    X_train_num_no_pv,
    X_train_cat
])

X_test_no_pv = np.hstack([
    X_test_num_no_pv,
    X_test_cat
])

# Train Random Forest with the same settings
rf_without_pagevalues = RandomForestClassifier(
    n_estimators=100,
    max_depth=10,
    random_state=42,
    class_weight="balanced",
    n_jobs=-1
)

rf_without_pagevalues.fit(X_train_no_pv, Y_train)

y_pred_rf_no_pv = rf_without_pagevalues.predict(X_test_no_pv)

print("Random Forest WITHOUT PageValues")
print("-" * 40)
print("Accuracy :", accuracy_score(Y_test, y_pred_rf_no_pv))
print("Precision:", precision_score(Y_test, y_pred_rf_no_pv, zero_division=0))
print("Recall   :", recall_score(Y_test, y_pred_rf_no_pv, zero_division=0))
print("F1 Score :", f1_score(Y_test, y_pred_rf_no_pv, zero_division=0))
print("Confusion Matrix:\n", confusion_matrix(Y_test, y_pred_rf_no_pv))
```

## Conclusion

In this project, I analyzed the Online Shoppers Purchasing Intention dataset to understand customer browsing behavior and predict whether a website session would result in a purchase.

I performed data cleaning, exploratory data analysis, visualization, feature preprocessing, and classification using K-Nearest Neighbors (KNN), Naive Bayes, Decision Tree, and Random Forest algorithms.

Among the models evaluated, the Random Forest classifier achieved the highest F1 score of       **65.97%**, with a recall of **82.72%** and an accuracy of **86.64%**. The Decision Tree also performed well, achieving a recall of **86.91%**.

Feature importance analysis identified **PageValues** as the most influential feature in the Random Forest model. An additional experiment excluding PageValues showed a substantial performance decrease, highlighting its importance to the model's predictions.

This project helped me understand classification, class imbalance, model evaluation, feature importance, and the trade-offs between precision and recall. Further improvements could include hyperparameter tuning, threshold optimization, and validating feature availability for real-time purchase prediction.
